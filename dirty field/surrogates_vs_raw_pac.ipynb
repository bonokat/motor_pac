```python
import pandas as pd
from pathlib import Path

# ============================================================
# LOAD DATA
# ============================================================
file_dir = Path(r"F:\research project\# study 1\eeg_data\set\Y group\roi_source_analysis\surrogate_analysis")
file_path = file_dir / "study1_surrogate_roi_summary.csv"
df = pd.read_csv(file_path)

# ============================================================
# 1. AVERAGE ACROSS ROIs WITHIN EACH PARTICIPANT
# ============================================================

df_sub = (
    df.groupby(
        ["sub", "condition", "task_stage"],
        as_index=False
    )
    .agg(
        original_pac=("original_pac", "mean"),
        surrogate_mean=("surrogate_mean", "mean"),
        surrogate_sd=("surrogate_sd", "mean"),
        surrogate_z=("surrogate_z", "mean"),
        delta_pac=("delta_pac", "mean")
    )
)

# ============================================================
# 2. CALCULATE GROUP-LEVEL MEANS
# ============================================================

df_group = (
    df_sub.groupby(
        ["condition", "task_stage"],
        as_index=False
    )
    .agg(
        original_pac=("original_pac", "mean"),
        surrogate_mean=("surrogate_mean", "mean"),
        surrogate_sd=("surrogate_sd", "mean"),
        surrogate_z=("surrogate_z", "mean"),
        delta_pac=("delta_pac", "mean"),
        n_subjects=("sub", "nunique")
    )
)

# ============================================================
# 3. DISPLAY RESULTS
# ============================================================

pd.set_option("display.float_format", lambda x: f"{x:.8f}")

print("\nCONDITION-LEVEL PAC SUMMARY")
print(df_group.to_string(index=False))

# ============================================================
# 4. SAVE RESULTS
# ============================================================

df_group.to_csv(
    "study1_pac_surrogate_condition_summary.csv",
    index=False
)
```
